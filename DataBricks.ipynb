```python
from pyspark.sql import SparkSession, functions as F
```

```python
spark = SparkSession.builder.master("local[*]").getOrCreate()
```

```python
sc=spark.sparkContext
```

```python
sc.textFile?
```

```python
!wget https://raw.githubusercontent.com/apache/spark/master/README.md
```

```python
text_file="README.md"
lines=sc.textFile(text_file)
```

```python
lines.count()
```

```python
lines.first()
```

```python
lines.take(10)
```

```python
lines.collect()
```

```python
lines.filter(lambda line : line != '').collect()
```

```python
lines.filter(lambda line : line != '').count()
```

```python
lines.map(lambda line:line.upper()).take(15)
```

```python
lines.filter(lambda line : line != '').map(lambda line:line.upper()).take(15)
```

```python
lines.map(lambda line:len(line)).take(10)
```

```python
from operator import add
lines.map(lambda line:len(line)).reduce(add)
```

```python
lines.filter(lambda line : line != '').flatMap(lambda line:line.split(' ')).take(10)
```

```python
lines.filter(lambda line : line != '').flatMap(lambda line:line.split(' ')).count()
```

```python
words=lines.filter(lambda line : line != '').flatMap(lambda line:line.split(' '))
words.map(lambda word :(word,1)).take(10)
```

```python
word_tupels=words.map(lambda word :(word.upper(),1))
word_tupels.reduceByKey(add).take(10)
```

```python
word_tupels.reduceByKey(add).count()
```

```python
word_tupels.reduceByKey(add).collect()
```
